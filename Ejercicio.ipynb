Serie Maclaurin

```python

import math

suma = 0

for n in range(1, 10000):
    suma += 4 * (-1)**(n+1) / (2*n-1)

    if abs(suma - math.pi) < 1e-3:
        print("Primer n:", n)
        print("Aproximación:", suma)
        print("Error:", abs(suma - math.pi))
        break
```

Fórmula de Machin

```python

import math

def P(n, x):
    suma = 0
    for i in range(1, n+1):
        suma += (-1)**(i+1) * x**(2*i-1) / (2*i-1)
    return suma

for n in range(1, 100):
    aproximacion = 4*(4*P(n, 1/5) - P(n, 1/239))
    error = abs(aproximacion - math.pi)

    if error < 1e-3:
        print("n =", n)
        print("Error =", error)
        break
```

Sucesión de Fibonacci

```python

import math

phi = (1 + math.sqrt(5)) / 2

a, b = 1, 1
i = 1

while True:
    r = b / a
    error = abs(phi - r) / phi

    if error < 1e-8:
        print("i =", i)
        print("Aproximación =", r)
        break

    a, b = b, a+b
    i += 1
```
